```python
import numpy as np
```

```python
#data type
arr =np.array([1,2,3,4.4,5])
print(arr,arr.dtype)
```

```python
#complex data type
arr=np.array([54 + 60j])
print(arr ,arr.dtype)
```

```python
#multi- dimensionsal array
arr2D = np.array([[2,3,4] ,[3,5,2],[4,4,2]])
sum_of_col=np.sum(arr2D,axis=0)
print(sum_of_col)
```
