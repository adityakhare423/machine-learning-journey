```python
# real data

#oscar winner movies
oscar_movies =["The Dark Knight",
               "the hurt locker",
               "the king's speech",
               "the artist",
               "argo"]
years = [2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

plt.plot(years,oscar_revenue)
plt.title("oscar movies revenue in eacch year")
plt.xlabel("years")
plt.ylabel(" revenue in ($M)")
              
```

```python
 # for non oscar movies
non_oscar_movies=[" slumdog millionaire", " avtar", "inception","hungo","lincoln"]
non_oscar_revenue =[ 378,2788,829,185,275]

#plt.plot(non_oscar_movies,non_oscar_revenue)
plt.plot(years,oscar_revenue ,  label="oscar movies")
plt.plot(years,non_oscar_revenue, label="non oscar movies")

plt.title("oscar movies vs non oscar revenue ")
plt.xlabel("years")
plt.ylabel(" revenue in ($M)")

plt.legend()

plt.style.use("dark_background")
plt.show()
```

```python
# bar chart
oscar_movies =["The Dark Knight",
               "the hurt locker",
               "the king's speech",
               "the artist",
               "argo"]
years = [2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

plt.bar(years,oscar_revenue)
plt.title("oscar movies revenue in eacch year")
plt.xlabel("years")
plt.ylabel(" revenue in ($M)")
              
for i in range(len(years)):
    plt.text(years[i],oscar_revenue[i]+10,str(oscar_revenue[i]),ha="center")
```

```python
import numpy as np

oscar_movies =["The Dark Knight",
               "the hurt locker",
               "the king's speech",
               "the artist",
               "argo"]
years = [2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

non_oscar_movies=[" slumdog millionaire", " avtar", "inception","hungo","lincoln"]
non_oscar_revenue =[ 378,2788,829,185,275]

x=np.arange(len(years))
width=0.4


plt.bar(x-width/2,oscar_revenue, width)
plt.bar(x+width/2,non_oscar_revenue, width)
plt.title("oscar vs non-oscar movies reveue")
plt.xlabel("years")
plt.ylabel(" revenue in ($M)")
```
