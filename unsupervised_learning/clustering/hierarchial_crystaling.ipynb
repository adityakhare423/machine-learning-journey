```python
from sklearn.datasets  import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris= load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
#scale
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
#visualise
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c="black")
```

```python
#scripy in the liberay to uses to make  the dendograprh 

from scipy.cluster.hierarchy import linkage, dendrogram

#linkage matrix for the store the data fro the deendogram
Z=linkage(X_scaled, method="ward")

#plot 
dendrogram(Z)
```

```python
#now makinng it better

from scipy.cluster.hierarchy import linkage, dendrogram

#linkage matrix for the store the data fro the deendogram
Z=linkage(X_scaled, method="ward")

#plot
plt.figure(figsize=(12,6))
dendrogram(Z)
plt.xlabel("sample")
plt.ylabel("distance")
plt.title("Dendrogram for the hierachial clustering")
```

```python
# now for the  clustering
from sklearn.cluster import AgglomerativeClustering

agg = AgglomerativeClustering(n_clusters=3)
labels = agg.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c=labels)
```
