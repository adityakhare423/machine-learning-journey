```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
titanic=sns.load_dataset("titanic")
```

```python
titanic.head()
```

```python
features=["pclass", "sex","fare","embarked", "age"]
target=["survived"]
```

```python
from sklearn.impute import SimpleImputer

median_imputer=SimpleImputer(strategy="median")
titanic[["age"]]=median_imputer.fit_transform(titanic[["age"]])


imputer_frequent=SimpleImputer(strategy="most_frequent")
titanic[["embarked"]]= imputer_frequent.fit_transform(titanic[["embarked"]])
```

```python
titanic[["age"]].head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

titanic["sex"]=le.fit_transform(titanic["sex"])
# from sklearn.preprocessing import OneHotEncoder
# ohe = OneHotEncoder(sparse_output=False)

titanic["embarked"] = le.fit_transform(titanic["embarked"])
```

```python
X=titanic[features]
y=titanic["survived"]


#trian test split
X_train, X_test, y_train, y_test = train_test_split(
    X,y,  test_size=0.3, random_state=42
)
```

```python
model = DecisionTreeClassifier(max_depth=5)
model.fit(X_train,y_train)
```

```python
y_pred_test=model.predict(X_test)
y_pred_train=model.predict(X_train)
```

```python
print("train acc",accuracy_score(y_train,y_pred_train)*100)
print("test acc",accuracy_score(y_test,y_pred_test)*100)    
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(18,10))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Died","Survied"],
    filled = True,
    # max_depth=3
)
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
random_for=RandomForestClassifier(
    n_estimators=501,
    oob_score=True,
    max_depth=5
)


random_for.fit(X_train,y_train)

y_pred=random_for.predict(X_test)

print("oob score",random_for.oob_score_*100)
print("testing accuracy" ,accuracy_score(y_test,y_pred)*100)
```
